```python
from google.oauth2 import service_account

credentials = service_account.Credentials.from_service_account_file(
    'service.json',
    scopes=['https://www.googleapis.com/auth/cloud-platform']
)
```

```python
from google import genai

genai_client = genai.Client(
    enterprise=True,
    location='global',
    project=credentials.project_id,
    credentials=credentials
)
```

```python
from time import time

with open('alj-0000.jpg', 'rb') as f:
    data = f.read()

contents = [
    {
        'role': 'user',
        'parts': [
            {'text': '車牌辨識'},
            {'inline_data': {'data': data, 'mime_type': 'image/jpeg'}}
        ]
    }
]

config = {
    'max_output_tokens': 2,
    #'thinking_config': {'thinking_level': 'minimal'},
    'thinking_config': {'thinking_level': 'low'}
}

start = time()
r = genai_client.models.generate_content(
    #model='gemini-3.5-flash-lite',
    model='gemini-3.8-flash',
    contents=contents,
    config=config
)
end = time()

if r.candidates[0].finish_reason == 'STOP':
    print(r.text)
else:
    print(r.candidates[0].finish_reason)
print(f'{end-start:.4f}s elapsed')
print(r.usage_metadata)
```
